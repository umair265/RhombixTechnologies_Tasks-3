```python
import pandas as pd

df = pd.read_csv('/content/Pakistan Largest Ecommerce Dataset.csv')

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
df = df.dropna(how='all')

print("Dataset shape after removing empty rows:", df.shape)
```

```python
df = df.dropna(axis=1, how='all')

print("Shape after removing empty columns:", df.shape)
```

```python
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

print(missing_percent.sort_values(ascending=False))
```

```python
df['sales_commission_code'] = df['sales_commission_code'].fillna('Unknown')

df = df.dropna(subset=[
    'category_name_1',
    'sku',
    'status',
    'Customer Since',
    'Customer ID'
])

print(df.isnull().sum())
```

```python
df['created_at'] = pd.to_datetime(
    df['created_at'], errors='coerce'
)

print(df['created_at'].dtype)
```

```python
df['Order_Month'] = df['created_at'].dt.month
df['Order_Year'] = df['created_at'].dt.year

df[['created_at', 'Order_Month', 'Order_Year']].head()
```

```python
df['Order_Month'] = df['created_at'].dt.month
df['Order_Year'] = df['created_at'].dt.year

df[['created_at', 'Order_Month', 'Order_Year']].head()
```

```python
df['Total_Sales'] = df['price'] * df['qty_ordered']

df[['price', 'qty_ordered', 'Total_Sales']].head()
```

```python
import numpy as np

df['Discount_Percentage'] = (
    df['discount_amount'] /
    df['Total_Sales'].replace(0, np.nan)
) * 100

df[['discount_amount', 'Total_Sales',
    'Discount_Percentage']].head()
```

```python
df['Customer_Lifetime_Value'] = (
    df.groupby('Customer ID')['Total_Sales']
    .transform('sum')
)

df[['Customer ID', 'Total_Sales',
    'Customer_Lifetime_Value']].head()
```

```python
print("Dataset Shape:", df.shape)

print("\nNew Features:")
print(df[['Order_Month',
          'Order_Year',
          'Total_Sales',
          'Discount_Percentage',
          'Customer_Lifetime_Value']].head())
```

```python
df.to_csv('Pakistan_Ecommerce_Cleaned.csv', index=False)

print("Dataset saved successfully!")
```
